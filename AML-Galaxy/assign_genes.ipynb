```python
data75 = EnsemblRelease(75)
```

```python
all_genes = []
for row in df.iterrows():
    chr = row[1]["chr"].replace("chr", "")
    s_pos = int(row[1]["start"])
    gene_names = data75.gene_names_at_locus(contig=chr, position=s_pos)
    all_genes.append(",".join(gene_names))

df["genes"] = all_genes
df
```

```python
df[df["genes"] == "SEPT9"]
```

```python
chr_SEPT9 = "chr17"
start_SEPT9 = 75277581 #75277618
end_SEPT9 = 75496678 #75494740
# chr17:77,280,569-77,500,596
# chr17:75277581-75496678 https://www.genecards.org/cgi-bin/carddisp.pl?gene=SEPTIN9

df_SEPT9= df[(df["chr"] == chr_SEPT9) & (df["start"] > start_SEPT9) & (df["end"] < end_SEPT9)]
df_SEPT9 = df_SEPT9.sort_values(by=["start"])
df_SEPT9
```
